## Run sod shock tube test for different resolution

```python
import os
import sys
import shutil
import numpy as np

output_dir = "/home/watanabe/Projects/outputs"
sod_dir = "/home/watanabe/Projects/outputs/Sod_CRinjection"
```

```python
# list of configurations 
amr_levels = [4,5,6,7]
```

For each configuration, copy flash.par and modify pleft, tend, and spacing for saving plot files

```python
for i, amr_lvl in enumerate(amr_levels):
    # if i != 0:
    #     continue
    # make directory
    sod_lvl_dir = os.path.join(output_dir, f"Sod_CRinjection_lvl{amr_lvl:.1f}")
    if not os.path.exists(sod_lvl_dir):
        os.mkdir(sod_lvl_dir)

    # copy flash4 & flash.par to this directory
    shutil.copy(os.path.join(sod_dir, "flash4"), sod_lvl_dir)
    shutil.copy(os.path.join(sod_dir, "flash.par"), sod_lvl_dir)

    # open flash.par
    flashpar_file = os.path.join(sod_lvl_dir, "flash.par")
    with open(flashpar_file, "r") as f:
        par_lines = f.readlines()

    # modify amr level
    for lvl_idx in [115, 116]:
        print(par_lines[lvl_idx])
        lvl_min_strlist = par_lines[lvl_idx].split(" ")
        lvl_min_strlist[-1] = f"{amr_lvl}\n"
        par_lines[lvl_idx] = " ".join(lvl_min_strlist)
        print(par_lines[lvl_idx])

    # print(par_lines[8])

    # finally write to flash.par
    with open(flashpar_file, "w") as f:
        f.writelines(par_lines)
```

```python
bash_tmpl = """#!/bin/bash -l

"""

for i, lvl in enumerate(amr_levels):
    bash_tmpl += f"echo lvl{lvl:.1f}\n"
    bash_tmpl += f"cd /home/watanabe/Projects/outputs/Sod_CRinjection_lvl{lvl:.1f}\n"
    bash_tmpl += f"rm -rf *sod* *.dump *.log\n"
    bash_tmpl += f"ls /home/watanabe/Projects/outputs/Sod_CRinjection_lvl{lvl:.1f}\n"
    bash_tmpl += f"mpirun -np 6 ./flash4 > /home/watanabe/Projects/outputs/Sod_CRinjection_lvl{lvl:.1f}/stdout.log\n"
    bash_tmpl += f"ls /home/watanabe/Projects/outputs/Sod_CRinjection_lvl{lvl:.1f}\n"

print(bash_tmpl)
with open("run_sod_amrlvls.sh", "w") as f:
    f.write(bash_tmpl)
```
