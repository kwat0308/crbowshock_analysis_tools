```python
import numpy as np
import astropy.units as u
```

```python
def calculate_outer_radius(Lwind, rho0, time):
    """As in Weaver et al. 1977"""
    return (0.76 * (Lwind / rho0)**0.2 * time**0.6).to(u.pc)

def calculate_dynamic_length(vstar, time):
    return (vstar * time).to(u.pc)
```

```python
# fixed values for both versions
Lwind = 6e34 * u.erg / u.s
rho0 = 2e-25 * u.g / u.cm**3
```

Previous version

```python
rad_v30 = calculate_outer_radius(Lwind, rho0, time=180 * u.kyr)
Ldyn_v30 = calculate_dynamic_length(vstar=30 * u.km / u.s, time=180 * u.kyr)

print(f"Outer Radius: {rad_v30:.2f}, Length scale: {Ldyn_v30:.2f}, Ratio (must be << 1): {rad_v30 / Ldyn_v30 :.2f}")
```

New verfsionsion

```python
rad_v40 = calculate_outer_radius(Lwind, rho0, time=500 * u.kyr)
Ldyn_v40 = calculate_dynamic_length(vstar=40 * u.km / u.s, time=500 * u.kyr)

print(f"Outer Radius: {rad_v40:.2f}, Length scale: {Ldyn_v40:.2f}, Ratio (must be << 1): {rad_v40 / Ldyn_v40 :.2f}")
```
