```python
import matplotlib.pyplot as plt
import numpy as np

import astropy.units as u
```

```python
data_path = "/home/kwatanabe/Projects/cr-bowshocks/updated_outputs/ISM-v100-MDiff"
```

```python
datafile_0 = np.genfromtxt(f"{data_path}/starISM.dat", comments="#")
datafile_1 = np.genfromtxt(f"{data_path}/CRsources.dat", comments="#", skip_header=1)
```

```python
plt.semilogy((datafile_0[:,0] * u.s).to(u.kyr), datafile_0[:,5], ls="", marker="o", ms=2.0, label="total energy", color="gray")
plt.semilogy((datafile_0[:,0] * u.s).to(u.kyr), datafile_0[:,6], ls="", marker="o", ms=2.0, label="kinetic energy", color="r")
plt.semilogy((datafile_0[:,0] * u.s).to(u.kyr), datafile_0[:,7], ls="", marker="o", ms=2.0, label="internal energy", color="g")
plt.semilogy((datafile_1[:,0] * u.s).to(u.kyr), datafile_1[:,2], ls="", marker="o", ms=2.0, label="cr energy", color="b")
plt.xlabel("Time / s")
plt.ylabel("Total energy / erg")
plt.legend()
```
