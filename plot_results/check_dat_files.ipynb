```python
data_path = "/home/kwatanabe/Projects/cr-bowshocks/outputs_new"
```

```python
datafile_0 = np.loadtxt(f"{data_path}/BS_ISM_v30_MHD_test/starISM.dat", comments="#")
datafile_1 = np.loadtxt(f"{data_path}/Bowshock_test/starISM.dat", comments="#")
datafile_2 = np.loadtxt(f"{data_path}/BS_ISM_v30_MHD/starISM.dat", comments="#")
datafile_3 = np.loadtxt(f"{data_path}/r25BS_ISM_v30_MHD/starISM.dat", comments="#")
```
