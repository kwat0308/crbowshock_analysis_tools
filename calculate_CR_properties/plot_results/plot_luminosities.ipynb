```python
import h5py
import matplotlib.pyplot as plt
import astropy.units as u
import astropy.constants as const
import numpy as np
import json

import matplotlib.lines as mlines
```

```python
flash_output_dir = "/home/kwatanabe/Projects/cr-bowshocks/cr_bowshock_outputs"
fig_dir = "/home/kwatanabe/Projects/cr-bowshocks/figures/emission"
data_dir = "/home/kwatanabe/Projects/cr-bowshocks/crbowshock_analysis_tools/data"

bowshock_config = json.load(open("../../config/bowshock_config.json", "r"))["config"]
```

## Gamma ray luminosity

```python
fig_tseries, ax_tseries = plt.subplots(figsize=(10, 6))

sim_handles=[]
for i, sim_name in enumerate(list(bowshock_config.keys())):
    # write data to h5py file
    with h5py.File(f"{data_dir}/emission_data_gamma.h5", "r") as pf:
        if sim_name not in list(pf.keys()):
            continue
        times = pf[sim_name]["times"][()]
        gamma_lum = pf[sim_name]["gamma_lum"][()] * u.erg * u.s**-1

    # remove outliers
    times = times[gamma_lum > 10 * u.erg * u.s**-1]
    gamma_lum = gamma_lum[gamma_lum > 10 * u.erg * u.s**-1]

    ax_tseries.semilogy(
        times[np.argsort(times)],
        gamma_lum[np.argsort(times)],
        color=bowshock_config[sim_name]["color"],
        label=sim_name,
        lw=2.0
    )
    sim_handles.append(
        mlines.Line2D(
            [0],
            [0],
            lw=2.0,
            color=bowshock_config[sim_name]["color"],
            label=bowshock_config[sim_name]["label"],
        )
    )

# plot data also
data_handles=[]
with h5py.File(f"{data_dir}/S14_data.h5", "r") as pf:
    for l, (bs_name, vd) in enumerate(list(pf.items())):
        d, d_err = vd["D"][()] * u.kpc
        flux = vd["flux"][()] * 1e-12 * (u.erg * u.s**-1 * u.cm**-2)
        flux_err = 0.05 * flux  # 5% calibration uncertainty

        # calcualte luminosity
        gamma_lum_data = (4 * np.pi * d**2 * flux).to(u.erg * u.s**-1)
        gamma_lum_data_err = (
            4 * np.pi * np.sqrt((2 * d * flux * d_err) ** 2 + (d**2 * flux_err) ** 2)
        )

        mean_gamma_lum_data = np.mean(gamma_lum_data)
        mean_gamma_lum_data_err = np.sqrt(np.sum(gamma_lum_data_err**2))

        ax_tseries.axhline(
            mean_gamma_lum_data.value, color="k", ls="--", alpha=1.0, label="BD+43° 3654", lw=2.0
        )
        ax_tseries.fill_between(
            np.linspace(-20, 200, 100),
            # (mean_gamma_lum_data - mean_gamma_lum_data_err).value,
            (mean_gamma_lum_data - mean_gamma_lum_data_err).value,
            (mean_gamma_lum_data).value,
            color="k",
            alpha=0.15,
            zorder=0
        )

        data_handles.append(mlines.Line2D(
            [0],
            [0],
            lw=2.0,
            color="k",
            label="BD+43° 3654 (Schulz+2014)",
            ls="--"
        ))


ax_tseries.set_xlabel("Time / kyr", fontsize=20)
ax_tseries.set_ylabel(r"$L_\gamma$ / erg s$^{-1}$", fontsize=20)

# limits
ax_tseries.set_xlim(xmin=0, xmax=180)
ax_tseries.set_ylim(ymin=2e31, ymax=6e33)

# ticks
ax_tseries.tick_params(
    axis="both",
    direction="in",
    which="major",
    size=8,
    labelsize=18,
    right=True,
    top=True,
)
ax_tseries.tick_params(
    axis="both",
    direction="in",
    which="minor",
    size=6,
    labelsize=16,
    right=True,
    top=True,
)

leg1 = fig_tseries.legend(
    handles=sim_handles,
    prop={"size": 16},
    loc="upper left",
    bbox_to_anchor=[0.15, 0.87],
    frameon=False
)
fig_tseries.add_artist(leg1)
fig_tseries.legend(
    handles=data_handles,
    prop={"size": 16},
    loc="upper left",
    bbox_to_anchor=[0.39, 0.87],
    frameon=False
)

# ax_tseries.legend(prop={"size": 15}, loc="upper left", bbox_to_anchor=[0.05, 0.98])

# save
fig_tseries.savefig(
    f"{fig_dir}/gamma/gamma_lum_timeseries.pdf", bbox_inches="tight", dpi=300
)
```

### Synchrotron luminosity per frequency vs time

```python
fig_tseries, ax_tseries = plt.subplots(figsize=(10, 6))

synch_freq = 1.5 * u.GHz  # choice of synchrotron frequency

for i, sim_name in enumerate(list(bowshock_config.keys())):
    print(sim_name)
    # if sim_name != "ISM-v30":
    #     continue
    if sim_name == "ISM-v30-MHD" or sim_name == "ISM-v60":
        continue

    # write data to h5py file
    with h5py.File(f"{data_dir}/emission_data_synch_timecorrected_ae4.2.h5", "r") as pf:
        print(pf[sim_name]["slices"].keys())
        times = pf[sim_name]["times"][()]
        synch_freq_grid = pf[sim_name]["synch_freq"][()] * u.GHz

        synch_lum = pf[sim_name]["synch_lum"][()] * (u.erg / u.s / u.Hz)

    synch_freq_idx = np.digitize(synch_freq, synch_freq_grid, right=True)
    synch_lum = synch_lum[:, synch_freq_idx]

    times = times[synch_lum > 0]
    synch_lum = synch_lum[synch_lum > 0]

    ax_tseries.semilogy(
        times, synch_lum, color=bowshock_config[sim_name]["color"], label=bowshock_config[sim_name]["label"], lw=2.0
    )

ax_tseries.set_xlabel("Time / kyr", fontsize=20)
ax_tseries.set_ylabel(
    rf"$L_\mathrm{{syn, {synch_freq:.1f}}}$ / erg s$^{{-1}}$ Hz$^{{-1}}$", fontsize=20
)

# ylimits
ax_tseries.set_xlim(xmin=0, xmax=180)
ax_tseries.set_ylim(ymin=5e19, ymax=5e21)

# ticks
ax_tseries.tick_params(
    axis="both",
    direction="in",
    which="major",
    size=8,
    labelsize=18,
    right=True,
    top=True,
)
ax_tseries.tick_params(
    axis="both",
    direction="in",
    which="minor",
    size=6,
    labelsize=16,
    right=True,
    top=True,
)

ax_tseries.legend(prop={"size": 16}, loc="upper left", frameon=False)

# save
fig_tseries.savefig(
    f"{fig_dir}/synch/synch_lum_timeseries.pdf", bbox_inches="tight", dpi=300
)
```

### Time-averaged Synchrotron luminosity vs frequency 

```python
fig_tseries, ax_tseries = plt.subplots(figsize=(10, 6))

sim_handles = []
for i, sim_name in enumerate(list(bowshock_config.keys())):
    # if sim_name != "ISM-v30":
    #     continue
    # if sim_name == "ISM-v100":
    #     continue
    if sim_name == "ISM-v30-MHD" or sim_name == "ISM-v60":
        continue

    # read data from h5py file
    with h5py.File(f"{data_dir}/emission_data_synch_timecorrected_ae4.2.h5", "r") as pf:
        times = pf[sim_name]["times"][()] * u.kyr
        synch_lum = pf[sim_name]["synch_lum"][()] * (u.erg / u.s / u.Hz)
        synch_freq_grid = pf[sim_name]["synch_freq"][()] * u.GHz

    # take the temporal average from 50 kyr or so

    print(synch_lum.shape)

    synch_lum = synch_lum[np.argsort(times)]
    times = np.sort(times)
    # tstart_idx = np.digitize(20 * u.kyr, times[:-1], right=True)
    synch_lum_tavg = np.mean(synch_lum, axis=0)
    # synch_lum_tavg = synch_lum[-1,:]

    a, b = np.polyfit(
        np.log10(synch_freq_grid.to_value(u.Hz)),
        np.log10((synch_freq_grid.to_value(u.Hz) * synch_lum_tavg.to_value(u.erg / u.s / u.Hz))),
        1,
    )
    print(a, b)

    ax_tseries.plot(
        np.log10(synch_freq_grid.to_value(u.Hz)),
        np.log10((synch_freq_grid.to(u.Hz) * synch_lum_tavg).to_value(u.erg / u.s)),
        color=bowshock_config[sim_name]["color"],
        label=sim_name,
        lw=2.0,
    )
    # ax_tseries.plot(
    #     np.log10(synch_freq_grid.to_value(u.Hz)),
    #     (-(2.2 - 3) / 2) * np.log10(synch_freq_grid.to_value(u.Hz)) + b + 4.5, 
    #     color="k",
    #     label=r"$\propto \nu^{-(s-3)/2}, \: s=2.2$",
    #     ls="--",
    #     lw=2.0,
    # )

    sim_handles.append(
        mlines.Line2D(
            [0],
            [0],
            lw=2.0,
            color=bowshock_config[sim_name]["color"],
            label=bowshock_config[sim_name]["label"],
        )
    )
# sim_handles.append(
#     mlines.Line2D(
#         [0],
#         [0],
#         lw=2.0,
#         color="k",
#         label=r"$\propto \nu^{-(s-3)/2}, \: s=2.2$",
#         ls="--",
#     )
# )

# plot data also
data_colors = ["k", "k"]
data_symbols = ["o", "d"]
data_labels = ["BD+43° 3654 (Moutzouri+2022)", "BD+60° 2522 (Moutzouri+2022)"]

data_handles = []
with h5py.File(f"{data_dir}/M22_data.h5", "r") as pf:
    for l, (bs_name, vd) in enumerate(list(pf.items())):
        synch_freq_data = vd["freq"][()] * u.GHz
        freq_err = 0.5 * u.GHz  # width of each sub-band
        d, d_err = vd["D"][()] * u.kpc
        flux = vd["flux"][()] * 1e-12 * (u.erg * u.s**-1 * u.cm**-2)
        # pure flux is 5% calibration uncertainty, but data is nu * flux so we have to include
        # frequency error for flux error calculation
        flux_err = flux * np.sqrt(0.05**2 + (freq_err / synch_freq_data) ** 2)

        # calcualte luminosity
        synch_lum_data = (4 * np.pi * d**2 * flux).to(u.erg * u.s**-1)
        synch_lum_data_err = (
            4 * np.pi * np.sqrt((2 * d * flux * d_err) ** 2 + (d**2 * flux_err) ** 2)
        ).to(u.erg * u.s**-1)

        log10_synch_lum_data = np.log10(synch_lum_data.value)
        log10_synch_lum_data_err = (
            synch_lum_data_err / (np.log(10) * synch_lum_data)
        ).value

        errorbar = ax_tseries.errorbar(
            np.log10(synch_freq_data.to_value(u.Hz)),
            log10_synch_lum_data,
            xerr=(freq_err / (np.log(10) * synch_freq_data)).value,
            yerr=log10_synch_lum_data_err,
            label=data_labels[l],
            marker=data_symbols[l],
            alpha=0.8,
            ls="",
            ms=6.0,
            color=data_colors[l],
            zorder=100
        )

        data_handles.append(errorbar)

freq_ls2355 = 944 * u.MHz
freq_ls2355_err = 288 * u.MHz
flux_density_ls2355 = 79.4 * u.mJy
flux_density_ls2355_err = 0.8 * u.mJy
d_ls2355 = 2.2 * u.kpc
d_ls2355_err = 0.1 * u.kpc
synch_lum_ls2355 = (4 * np.pi * d_ls2355**2 * flux_density_ls2355 * freq_ls2355).to(u.erg / u.s)
synch_lum_ls2355_err = (
    4 * np.pi * np.sqrt(
        (2 * d_ls2355 * freq_ls2355 * flux_density_ls2355 * d_ls2355_err) ** 2
        + (d_ls2355**2 * freq_ls2355 * flux_density_ls2355_err) ** 2
        + (d_ls2355**2 * flux_density_ls2355 * freq_ls2355_err) ** 2
    )
).to(u.erg / u.s)

log10_synch_lum_ls2355 = np.log10(synch_lum_ls2355.value)
log10_synch_lum_ls2355_err = (
    synch_lum_ls2355_err / (np.log(10) * synch_lum_ls2355)
).value

print(log10_synch_lum_ls2355, log10_synch_lum_ls2355_err)

errorbar = ax_tseries.errorbar(
    np.log10(freq_ls2355.to_value(u.Hz)),
    log10_synch_lum_ls2355,
    xerr=(freq_ls2355_err / (np.log(10) * freq_ls2355)).value,
    yerr=log10_synch_lum_ls2355_err,
    label="LS 2355 (van den Eijnden+2024)",
    marker="s",
    alpha=0.8,
    ls="",
    ms=6.0,
    color="k",
    zorder=100
)
data_handles.append(errorbar)

ax_tseries.set_xlabel(r"$\log_{10}(\nu$ / Hz)", fontsize=20)
ax_tseries.set_ylabel(
    r"$\log_{10}\left(\nu \langle L_{\nu} \rangle \: / \: \mathrm{erg} \, \mathrm{s}^{-1} \right)$",
    fontsize=20,
)


# ylimits
ax_tseries.set_xlim(8.8, 10.1)
ax_tseries.set_ylim(ymin=28.8, ymax=33)

# ticks
ax_tseries.tick_params(
    axis="both",
    direction="in",
    which="major",
    size=8,
    labelsize=18,
    right=True,
)
ax_tseries.tick_params(
    axis="both",
    direction="in",
    which="minor",
    size=6,
    labelsize=16,
    right=True,
)

# generate twin axis for photon energy
secax = ax_tseries.secondary_xaxis(
    "top",
    functions=(
        lambda x: np.log10((10**x * u.Hz * const.h.si).to_value(u.eV)),
        lambda x: np.log10((10**x * u.eV / const.h.si).to_value(u.Hz)),
    ),
)
secax.set_xlabel(
    r"$\log_{10}(E_\mathrm{syn} = h\, \nu$ / $\mathrm{eV})$", fontsize=20
)
secax.tick_params(axis="x", which="major", labelsize=18, size=8, direction="in")
secax.tick_params(axis="x", which="minor", labelsize=18, size=6, direction="in")

# ax_tseries.legend(prop={"size": 14}, loc="upper left")
leg1 = fig_tseries.legend(
    handles=sim_handles,
    prop={"size": 16},
    loc="upper left",
    bbox_to_anchor=[0.14, 0.87],
    frameon=False
)
fig_tseries.add_artist(leg1)
fig_tseries.legend(
    handles=data_handles,
    prop={"size": 16},
    loc="upper left",
    bbox_to_anchor=[0.39, 0.87],
    frameon=False
)

fig_tseries.savefig(f"{fig_dir}/synch/synch_lum_freq.pdf", bbox_inches="tight", dpi=300)
```
