```python
import h5py
```

synchrotron flux from Moutzouri 2022

```python
with h5py.File("../../data/M22_data.h5", "w") as f:
    bsbd43_gr = f.create_group("BSBD43")
    bsbd43_gr.create_dataset("D", data=[1.72, 0.03])
    bsbd43_gr.create_dataset("freq", data=[1.4, 2, 4.5, 5.5, 6.5, 7.5, 9.5])
    bsbd43_gr.create_dataset("flux", data=[0.0014, 0.014, 0.007, 0.0078, 0.0078, 0.005175, 0.00437])

    bsbd60_gr = f.create_group("BSBD60")
    bsbd60_gr.create_dataset("D", data=[3.0, 0.2])
    bsbd60_gr.create_dataset("freq", data=[4.5, 5.5, 6.5, 7.5, 8.5, 10.5])
    bsbd60_gr.create_dataset("flux", data=[0.037, 0.0385, 0.0432, 0.04485, 0.04335, 0.04515])
```
